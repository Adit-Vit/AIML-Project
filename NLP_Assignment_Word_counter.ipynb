Program 4 : Rewrite Sentences from the parse Tree

```python
import nltk
from nltk import CFG, ChartParser
# Define grammar
grammar = CFG.fromstring("""
S -> NP VP
NP -> PropN | Det Adj N | Det N
VP -> V NP
Det -> 'a' | 'the'
Adj -> 'skilled' | 'young'
N -> 'student' | 'programmer' | 'engineer'
V -> 'is'
PropN -> 'Aditya' """)
# Create parser
parser = ChartParser(grammar)
# Original sentence
sentence = "Aditya is a skilled programmer"
# Tokenize sentence
tokens = sentence.split()
```

```python
# Generate parse tree
trees = list(parser.parse(tokens))
if trees:
    tree = trees[0]

    print("PARSE TREE")
    print("----------")
    tree.pretty_print()

    # Extract words from parse tree
    words = tree.leaves()

    # Rewrite sentence
    rewritten_sentence = " ".join(words)

    print("\nOriginal Sentence:")
    print(sentence)

    print("\nRewritten Sentence:")
    print(rewritten_sentence)
else:
    print("No parse tree found.")
```
