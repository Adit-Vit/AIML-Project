[4] Building a Context-Free Grammar and Parsing

```python
from nltk import CFG
from nltk.parse.generate import generate


# Define a simple CFG
cfg = CFG.fromstring("""
  S -> NP VP
  NP -> Det N
  VP -> V NP
  Det -> 'to' | 'the'
  N -> 'He' | 'school'
  V -> 'went'
""")


print("Generated Sentences:")
for sentence in generate(cfg, n=10):
    print(' '.join(sentence))
```

[5] Constituency Parsing using a Chart Parser

```python
from nltk import CFG
from nltk.parse.chart import ChartParser


# Define a simple CFG
cfg = CFG.fromstring("""
  S -> NP VP | Modal NP VP | NP VP PP
  NP -> Det N | Pronoun
  VP -> V NP | V PP | V Adv
  PP -> P NP
  Det -> 'A' | 'the' | 'a' | 'The'
  N -> 'hen' | 'cat' | 'rat' | 'salt' | 'mouse'| 'lion' | 'zebra'
  V -> 'chased' | 'saw' | 'pass'
  Pronoun -> 'you'
  Modal -> 'could'
  Adv -> 'please'
  P -> '?'
""")


# Use the same grammar
parser = ChartParser(cfg)


sentence = ['The', 'lion', 'chased', 'a', 'zebra']
print("\nConstituency Parse Trees:")
for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```

[18] Natural Language in Programming (NL to SQL)

```python
def nl_to_sql(nl_query):
    if "employees" in nl_query.lower():
        return "SELECT * FROM employees;"
    elif "departments" in nl_query.lower():
        return "SELECT * FROM departments;"
    return "Query not understood."

print(nl_to_sql("Show me all employees"))
```

[19] Expert System in AI (Diagnosis)

```python
def expert_system(symptom):
    if symptom == "fever":
        return "Possible diagnosis: Flu"
    elif symptom == "chest pain":
        return "Possible diagnosis: Heart issue"
    else:
        return "Diagnosis unknown"

print(expert_system("fever"))

# Expert System for Medical Diagnosis


def get_yes_no_input(prompt):
    """Helper function to get Yes/No input from the user."""
    while True:
        response = input(prompt + " (yes/no): ").lower()
        if response in ["yes", "no"]:
            return response == "yes"
        print("Please enter 'yes' or 'no'.")


def diagnose():
    print("Medical Diagnosis Expert System")
    print("-" * 35)

    fever = get_yes_no_input("Do you have a fever?")
    cough = get_yes_no_input("Do you have a cough?")
    sore_throat = get_yes_no_input("Do you have a sore throat?")
    fatigue = get_yes_no_input("Are you experiencing fatigue?")
    headache = get_yes_no_input("Do you have a headache?")
    nausea = get_yes_no_input("Do you feel nauseous?")
    rash = get_yes_no_input("Do you have any skin rashes?")

    # Rules-based diagnosis
    if fever and cough and fatigue:
        print("\nPossible Diagnosis: Influenza (Flu)")
    elif fever and sore_throat and cough:
        print("\nPossible Diagnosis: Common Cold")
    elif headache and nausea and not fever:
        print("\nPossible Diagnosis: Migraine")
    elif fever and rash:
        print("\nPossible Diagnosis: Measles or Dengue")
    elif sore_throat and no_symptoms(fever, cough, fatigue, headache, nausea):
        print("\nPossible Diagnosis: Throat Infection")
    else:
        print("\nDiagnosis: Symptoms are unclear. Please consult a medical professional.")


def no_symptoms(*symptoms):
    return all(not s for s in symptoms)


# Run the expert system
if __name__ == "__main__":
    diagnose()
```

[20] Recommendation System in AI

```python
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.feature_extraction.text import CountVectorizer

movies = ["Action adventure with explosions",
          "Romantic love story",
          "Explosions and car chases",
          "Love and heartbreak drama"]

user_likes = "I like action and adventure"

vectorizer = CountVectorizer().fit_transform(movies + [user_likes])
similarity = cosine_similarity(vectorizer[-1], vectorizer[:-1])
recommended = movies[similarity.argmax()]
print("Recommended:", recommended)
```
