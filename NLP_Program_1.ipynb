```python

# Step 2: Import and download required tokenizer models
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# Step 3: Use sent_tokenize
from nltk.tokenize import sent_tokenize
```

```python
text1 = "Artificial intelligence (AI) is the capability of computational systems to perform tasks typically associated with human intelligence, such as learning, reasoning, problem-solving, perception, and decision-making. It is a field of research in computer science that develops and studies methods and software that enable machines to perceive their environment and use learning and intelligence to take actions that maximize their chances of achieving defined goals"
```

```python
# Tokenizing into sentences
#sentences = sent_tokenize(text)
sentences = sent_tokenize(text1)
```

```python
# Print the result
print(sentences)
```

[2] Part of Speech [PoS] Tagging

```python
import nltk
nltk.download('punkt') # Pre-trained unsupervised model for tokenizing text into sentences or words.
nltk.download('averaged_perceptron_tagger') # Averaged Perceptron algorithm, a type of supervised machine learning model.
nltk.download('averaged_perceptron_tagger_eng') #
nltk.download('punkt_tab')
```

```python
# Sample sentence
#text1 = "The quick brown fox jumps over the lazy dog."
#text = "Artificial intelligence (AI) is the capability of computational systems to perform tasks typically associated with human intelligence, such as learning, reasoning, problem-solving, perception, and decision-making. It is a field of research in computer science that develops and studies methods and software that enable machines to perceive their environment and use learning and intelligence to take actions that maximize their chances of achieving defined goals"
text2 = "Hello! How are you doing today? I had a breakfast! I hope you're having a great day."
# Tokenize words
tokens = nltk.word_tokenize(text2)
```

```python
# POS Tagging
tagged = nltk.pos_tag(tokens)
print(tagged)
```
